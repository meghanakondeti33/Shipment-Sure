```python
# ============================================================
# FEATURE ENGINEERING
# ============================================================

model_df = df.copy()

# Cost-to-Weight Ratio
model_df["Cost_to_Weight_Ratio"] = (
    model_df["Cost_of_the_Product"]
    /
    model_df["Weight_in_gms"].replace(0, np.nan)
)

# Replace infinite values
model_df.replace(
    [np.inf, -np.inf],
    np.nan,
    inplace=True
)

# Fill missing engineered feature
model_df["Cost_to_Weight_Ratio"] = (
    model_df["Cost_to_Weight_Ratio"]
    .fillna(
        model_df["Cost_to_Weight_Ratio"].median()
    )
)

print("Feature engineering completed.")

display(
    model_df.head()
)
```

```python
# ============================================================
# FEATURES AND TARGET
# ============================================================

TARGET = "Reached.on.Time_Y.N"

# ID is only an identifier
X = model_df.drop(
    columns=[
        "ID",
        TARGET
    ]
)

y = model_df[TARGET]

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nFeatures:")
print(X.columns.tolist())

print("\nTarget:")
print(y.name)
```

```python
# ============================================================
# FEATURE TYPES
# ============================================================

categorical_features = X.select_dtypes(
    include=["object"]
).columns.tolist()

numerical_features = X.select_dtypes(
    include=[np.number]
).columns.tolist()

print("Categorical features:")
print(categorical_features)

print("\nNumerical features:")
print(numerical_features)
```

```python
# ============================================================
# TRAIN / VALIDATION / TEST SPLIT
# ============================================================

# 80% training pool
# 20% untouched test set

X_train_full, X_test, y_train_full, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# From training pool:
# 80% actual training
# 20% validation

X_train, X_val, y_train, y_val = train_test_split(
    X_train_full,
    y_train_full,
    test_size=0.20,
    random_state=42,
    stratify=y_train_full
)

print("Training:", X_train.shape)
print("Validation:", X_val.shape)
print("Test:", X_test.shape)

print("\nTarget distribution")

print("\nTraining:")
print(
    y_train.value_counts(normalize=True)
)

print("\nValidation:")
print(
    y_val.value_counts(normalize=True)
)

print("\nTest:")
print(
    y_test.value_counts(normalize=True)
)
```

```python
# ============================================================
# PREPROCESSING PIPELINE
# ============================================================

# Numerical preprocessing
numeric_transformer = SklearnPipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)

# Categorical preprocessing
categorical_transformer = SklearnPipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)

# Combine numerical + categorical processing
preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            numeric_transformer,
            numerical_features
        ),
        (
            "cat",
            categorical_transformer,
            categorical_features
        )
    ]
)

print("Preprocessing pipeline created.")
```

```python
# ============================================================
# MACHINE LEARNING MODELS
# ============================================================

models = {

    "Logistic Regression": LogisticRegression(
        C=0.5,
        max_iter=1000,
        random_state=42
    ),

    "Decision Tree": DecisionTreeClassifier(
        max_depth=8,
        min_samples_split=10,
        min_samples_leaf=5,
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        max_depth=10,
        min_samples_split=10,
        min_samples_leaf=5,
        random_state=42,
        n_jobs=-1
    ),

    "Naive Bayes": GaussianNB(),

    "KNN": KNeighborsClassifier(
        n_neighbors=9
    ),

    "SVM": SVC(
        C=0.5,
        kernel="rbf",
        probability=True,
        random_state=42
    ),

    "XGBoost": XGBClassifier(
        learning_rate=0.05,
        n_estimators=500,
        max_depth=6,
        subsample=0.8,
        colsample_bytree=0.8,
        reg_lambda=0.5,
        eval_metric="logloss",
        random_state=42
    ),

    "LightGBM": LGBMClassifier(
        learning_rate=0.05,
        n_estimators=500,
        max_depth=6,
        num_leaves=20,
        subsample=0.8,
        colsample_bytree=0.8,
        reg_alpha=0.2,
        reg_lambda=0.5,
        random_state=42,
        verbosity=-1
    ),

    "CatBoost": CatBoostClassifier(
        iterations=500,
        learning_rate=0.05,
        depth=6,
        l2_leaf_reg=5,
        verbose=False,
        random_state=42
    )
}

print("Models defined:")
for name in models:
    print("-", name)
```

```python
# ============================================================
# TRAIN ALL MODELS
# ============================================================

results = []

trained_pipelines = {}

for name, model in models.items():

    print("\n" + "=" * 70)
    print("Training:", name)
    print("=" * 70)

    # Complete pipeline:
    #
    # Raw data
    #     ↓
    # Preprocessing
    #     ↓
    # SMOTE
    #     ↓
    # Model

    pipeline = ImbPipeline(
        steps=[
            (
                "preprocessor",
                preprocessor
            ),

            (
                "smote",
                SMOTE(
                    random_state=42
                )
            ),

            (
                "model",
                model
            )
        ]
    )

    # Train
    pipeline.fit(
        X_train,
        y_train
    )

    # Validation predictions
    y_pred = pipeline.predict(
        X_val
    )

    y_prob = pipeline.predict_proba(
        X_val
    )[:, 1]

    # Metrics
    accuracy = accuracy_score(
        y_val,
        y_pred
    )

    precision = precision_score(
        y_val,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_val,
        y_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_val,
        y_pred,
        zero_division=0
    )

    roc_auc = roc_auc_score(
        y_val,
        y_prob
    )

    # Store
    results.append({
        "Model": name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1,
        "ROC-AUC": roc_auc
    })

    trained_pipelines[name] = pipeline

    print(
        f"Accuracy : {accuracy:.4f}"
    )

    print(
        f"Precision: {precision:.4f}"
    )

    print(
        f"Recall   : {recall:.4f}"
    )

    print(
        f"F1 Score : {f1:.4f}"
    )

    print(
        f"ROC-AUC  : {roc_auc:.4f}"
    )
```

```python
# ============================================================
# MODEL COMPARISON
# ============================================================

comparison_df = pd.DataFrame(
    results
)

comparison_df = comparison_df.sort_values(
    by="ROC-AUC",
    ascending=False
).reset_index(drop=True)

display(
    comparison_df.round(4)
)
```

```python
# ============================================================
# VERIFY MODEL PREDICTIONS
# ============================================================

prediction_summary = []

for name, pipeline in trained_pipelines.items():

    predictions = pipeline.predict(
        X_val
    )

    unique_values, counts = np.unique(
        predictions,
        return_counts=True
    )

    prediction_summary.append({
        "Model": name,
        "Class 0 Predictions": counts[
            list(unique_values).index(0)
        ] if 0 in unique_values else 0,

        "Class 1 Predictions": counts[
            list(unique_values).index(1)
        ] if 1 in unique_values else 0
    })

prediction_summary_df = pd.DataFrame(
    prediction_summary
)

display(
    prediction_summary_df
)
```

```python
# ============================================================
# 10-FOLD STRATIFIED CROSS VALIDATION
# XGBOOST
# ============================================================

cv = StratifiedKFold(
    n_splits=10,
    shuffle=True,
    random_state=42
)

xgb_cv_pipeline = ImbPipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),

        (
            "smote",
            SMOTE(
                random_state=42
            )
        ),

        (
            "model",
            XGBClassifier(
                learning_rate=0.05,
                n_estimators=500,
                max_depth=6,
                subsample=0.8,
                colsample_bytree=0.8,
                reg_lambda=0.5,
                eval_metric="logloss",
                random_state=42
            )
        )
    ]
)

cv_scores = cross_val_score(
    xgb_cv_pipeline,
    X_train,
    y_train,
    cv=cv,
    scoring="accuracy",
    n_jobs=-1
)

print("10-Fold Cross Validation Results")
print("=" * 60)

print(
    "Fold scores:",
    np.round(cv_scores, 4)
)

print(
    f"\nMean Accuracy: "
    f"{cv_scores.mean():.4f}"
)

print(
    f"Standard Deviation: "
    f"{cv_scores.std():.4f}"
)
```

```python
# ============================================================
# SELECT FINAL MODEL
# ============================================================

final_xgb_pipeline = trained_pipelines[
    "XGBoost"
]

print(
    "XGBoost pipeline selected as final model."
)
```

```python
# ============================================================
# XGBOOST VALIDATION PERFORMANCE
# ============================================================

xgb_val_pred = final_xgb_pipeline.predict(
    X_val
)

xgb_val_prob = final_xgb_pipeline.predict_proba(
    X_val
)[:, 1]

print("=" * 60)
print("XGBOOST VALIDATION RESULTS")
print("=" * 60)

print(
    f"Accuracy : "
    f"{accuracy_score(y_val, xgb_val_pred):.4f}"
)

print(
    f"Precision: "
    f"{precision_score(y_val, xgb_val_pred, zero_division=0):.4f}"
)

print(
    f"Recall   : "
    f"{recall_score(y_val, xgb_val_pred, zero_division=0):.4f}"
)

print(
    f"F1 Score : "
    f"{f1_score(y_val, xgb_val_pred, zero_division=0):.4f}"
)

print(
    f"ROC-AUC  : "
    f"{roc_auc_score(y_val, xgb_val_prob):.4f}"
)
```

```python
# ============================================================
# XGBOOST THRESHOLD OPTIMIZATION
# ============================================================

precision, recall, thresholds = precision_recall_curve(
    y_val,
    xgb_val_prob
)

# Calculate F1 score for each threshold
f1_scores = (
    2
    * precision[:-1]
    * recall[:-1]
    /
    (
        precision[:-1]
        + recall[:-1]
        + 1e-12
    )
)

best_idx = np.argmax(
    f1_scores
)

best_threshold = thresholds[
    best_idx
]

best_validation_f1 = f1_scores[
    best_idx
]

print(
    f"Best threshold: "
    f"{best_threshold:.4f}"
)

print(
    f"Best validation F1: "
    f"{best_validation_f1:.4f}"
)
```

```python
# ============================================================
# FINAL TEST SET EVALUATION
# ============================================================

# Test probabilities
xgb_test_prob = final_xgb_pipeline.predict_proba(
    X_test
)[:, 1]

# Apply optimized threshold
xgb_test_pred = (
    xgb_test_prob >= best_threshold
).astype(int)

# Metrics
test_accuracy = accuracy_score(
    y_test,
    xgb_test_pred
)

test_precision = precision_score(
    y_test,
    xgb_test_pred,
    zero_division=0
)

test_recall = recall_score(
    y_test,
    xgb_test_pred,
    zero_division=0
)

test_f1 = f1_score(
    y_test,
    xgb_test_pred,
    zero_division=0
)

test_roc_auc = roc_auc_score(
    y_test,
    xgb_test_prob
)

print("=" * 70)
print("FINAL XGBOOST TEST RESULTS")
print("=" * 70)

print(
    f"Accuracy : {test_accuracy:.4f}"
)

print(
    f"Precision: {test_precision:.4f}"
)

print(
    f"Recall   : {test_recall:.4f}"
)

print(
    f"F1 Score : {test_f1:.4f}"
)

print(
    f"ROC-AUC  : {test_roc_auc:.4f}"
)
```

```python
# ============================================================
# CLASSIFICATION REPORT
# ============================================================

print(
    classification_report(
        y_test,
        xgb_test_pred,
        target_names=[
            "Not On Time",
            "On Time"
        ],
        zero_division=0
    )
)
```

```python
# ============================================================
# ROC CURVE
# ============================================================

fpr, tpr, _ = roc_curve(
    y_test,
    xgb_test_prob
)

roc_auc_value = auc(
    fpr,
    tpr
)

plt.figure(figsize=(7, 6))

plt.plot(
    fpr,
    tpr,
    label=f"XGBoost AUC = {roc_auc_value:.4f}"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--"
)

plt.xlabel(
    "False Positive Rate"
)

plt.ylabel(
    "True Positive Rate"
)

plt.title(
    "XGBoost ROC Curve"
)

plt.legend(
    loc="lower right"
)

plt.tight_layout()
plt.show()
```
